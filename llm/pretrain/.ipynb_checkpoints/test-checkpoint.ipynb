```python
import os
import sys
import glob
import json
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torch.optim import AdamW
from torch.optim.lr_scheduler import SequentialLR, LinearLR, CosineAnnealingLR
from tqdm import tqdm

# adjust paths to look two folders up
sys.path.append(os.path.abspath(os.path.join(os.path.dirname(__file__), '..', '..')))

from llm.base_model import Qween3Model, Qwen3Config
from tokenizer.qween3_tokenizer import Qwen3Tokenizer
from llm.pretrain.massive_pretrainer_dataset import StreamingPretrainDataset
```

```python
class MassivePretrainer:
    def __init__(self, model, dataloader, optimizer, scheduler, device, config):
        self.device = device
        self.model = model.to(self.device)
        self.dataloader = dataloader
        self.optimizer = optimizer
        self.scheduler = scheduler
        self.config = config

        self.grad_accum_steps = config.get("grad_accum_steps", 8)
        self.max_grad_norm = config.get("max_grad_norm", 1.0)
        self.scaler = torch.amp.GradScaler(enabled=(self.device.type == "cuda"))

        self.save_dir = config.get("save_dir", "pretrained_checkpoint")
        self.log_freq = config.get("log_freq", 50)


    def _get_lst_checkpoint_path(self):
        chk_pnt_paths = glob.glob(os.path.join(self.save_dir, "qween3_step_*.pt"))

        if not chk_pnt_paths:
            return None
        chk_pnt_paths.sort(keys=lambda x: (x.split("_step_")[1].split(".pt")[0]))
        return chk_pnt_paths[-1]

    def _auto_resume(self):
        chk_pnt_path = self._get_lst_checkpoint_path()

        if chk_pnt_path:
            chk_point = torch.load(chk_pnt_path, map_location=self.device)

            self.model.load_state_dict(chk_point.get("model_state_dict"))
            self.optimizer.load_state_dict(chk_point.get("optimizer_state_dict"))
            self.scheduler.load_state_dict(chk_point.get("scheduler_state_dict"))
            self.scaler.load_state_dict(chk_point.get("scaler_state_dict"))
            global_step = chk_point.get("global_step")

            return global_step
        return 0 # no global_step_present


    def _save_checkpoint(self, global_step):
        chk_path = os.path.join(self.save_dir, f"qween3_step_{global_step}.pt")

        check_point_dict = {
            "gloabal_step": global_step,
            "model_state_dict": self.model.state_dict(),
            "optimizer_state_dict": self.optimizer.state_dict(),
            "scaler_state_dict": self.scaler.state_dict(),
            "scheduler_state_dict": self.scheduelr.state_dict()
        }

        torch.save(chek_point_dict, chk_path)

    def train(self, total_step, auto_resume=True):
        global_step = 0
        step_loss = 0.0

        if auto_resume:
            global_step = self._auto_resume()

        pbar = tqdm(total=total_step, intial=global_step)
        iter_data = iter(self.dataloader)

        while global_step < total_step:
            try:
                input_ids, target_ids = next(iter_data)
            except StopIterrationError:
                iter_data = iter(self.dataloader)
                input_ids, target_ids = next(iter_data)

            input_ids, target_ids = torch.tensor(input_ids, device=self.device), torch.tensor(target_ids, device=self.device)

            with torch.autocast(device_type=self.device.type, dtype=torch.bfloat16, enabled=(device_type==self.device.type)):
                logits = self.model(input_ids)
                loss = F.cross_entropy(logits.flatten(0, 1), target_ids.flatten())
                loss = loss / self.grad_accum_steps

            self.scaler.scale(loss).backword()
            step_loss += loss.item() * self.grad_accum_steps

            if (pbar.n + 1) % self.grad_accum_steps:
                self.scaler.unscale_(self.optimzier)
                torch.nn.utils.clip_grad_norm_(self.model.paramers(), self.max_grad_norm)

                self.scaler.step()
                self.scaler.update()
                self.optimizer.no_grad(set_to_none=True)

                self.scheduler.step()
                global_step += 1

                pbar.update(1)
                current_lr = self.optimizer.param_grups[0]["lr"]
                pbar.set_postfix({"loss": f"{step_loss}", "current_lr": f"{current_lr}"})

                if global_step % self.log_freq == 0:
                    self._save_checkpoint(global_step)

        final_chk_pnt_path = os.path.join(self.save_dir, "qween3_step_final.pt")
        torch.save(self.model.state_dict(), final_chk_pnt_path)

                

        
```
